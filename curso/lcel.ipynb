La Cadena Simple (Simple Chain)
El concepto clave de LCEL es el uso del operador |. En Python tradicional, esto significa "un 'O' lógico", pero en LangChain significa: "Toma la salida del lado izquierdo y pásala como entrada al lado derecho".

```python
from dotenv import load_dotenv
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()
```

```python
# 1. Definimos el Prompt (Plantilla de texto)
prompt = ChatPromptTemplate.from_template("Cuéntame un chiste corto sobre {topic}")
```

```python
# 2. Definimos el Modelo moderno (Gemini)
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
# 3. Definimos el Parser (Limpia la respuesta para que sea solo texto plano)
output_parser = StrOutputParser()
```

```python
# 4. LA MAGIA DE LCEL: Construimos la cadena (Chain)
# El Prompt pasa su texto al Modelo, y el Modelo le pasa su respuesta cruda al Parser.
chain = prompt | model | output_parser
```

```python
# 5. Invocamos la cadena pasando el parámetro obligatorio
resultado = chain.invoke({"topic": "osos"})
print(resultado)
```
