Control de Errores y Alternativas (Fallbacks)
Hacia el final, tu cuaderno muestra una característica espectacular de LCEL: los Fallbacks (Planes de respaldo). Si un modelo falla, se cae el servidor o arroja un error de formato, la cadena cambia automáticamente a un segundo modelo para que tu aplicación nunca deje de funcionar.

```python
from dotenv import load_dotenv
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import json
```

```python
load_dotenv()
```

```python
# Supongamos que configuramos un modelo con un nombre incorrecto (va a fallar por fuerza)
modelo_defectuoso = ChatGoogleGenerativeAI(model="gemini-modelo-que-no-existe")
```

```python
# Creamos nuestro modelo de respaldo (el confiable Gemini 2.5)
modelo_respaldo = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
# Le añadimos el fallback al defectuoso
# Traducido: "Intenta usar el defectuoso, si falla por cualquier motivo, usa el de respaldo"
cadena_segura = modelo_defectuoso.with_fallbacks([modelo_respaldo])
```

```python
# Probamos la ejecución
try:
    respuesta = cadena_segura.invoke("Escribe un poema de dos líneas")
    print("--- ¡Éxito gracias al Plan de Respaldo! ---")
    print(respuesta.content)
except Exception as e:
    print(f"La aplicación se rompió por completo: {e}")
```
